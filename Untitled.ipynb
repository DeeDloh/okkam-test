```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(
    "data.csv",
    sep=",",
    index_col=0

)
```

```python
print(df.shape)
print(df.info())
print(df.head(10))
```

```python
df
```

```python
df["Date"] = pd.to_datetime(
    df["Date"].astype(str),
    format="%Y%m%d",
    errors="coerce"
)

df["Weight"] = pd.to_numeric(
    df["Weight"].astype(str).str.replace(",", ".", regex=False),
    errors="coerce"
)

for col in ["respondent", "Sex", "Age"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
print(
    df["Weight"].describe(
        percentiles=[0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99]
    )
)
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    df["Weight"].dropna(),
    bins=50
)

plt.title("Распределение Weight")
plt.xlabel("Weight")
plt.ylabel("Количество наблюдений")
plt.show()
```

```python
print(
    df.groupby("Sex")["Weight"]
    .agg(["count", "mean", "median", "std"])
)

print(
    df.groupby(pd.cut(df["Age"], bins=[0, 18, 25, 35, 45, 60, 120]))
    ["Weight"].agg(["count", "mean"])
)
```

```python
print("Начало периода:", df["Date"].min())
print("Конец периода:", df["Date"].max())
print("Уникальных дат:", df["Date"].nunique())
```

```python
daily = df.groupby("Date").size()

daily.plot(figsize=(12, 5))

plt.title("Количество наблюдений по датам")
plt.xlabel("Дата")
plt.ylabel("Количество")
plt.grid(True)
plt.show()
```
